```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# More libraries/tools here
from catboost import CatBoostRegressor, CatBoostClassifier
```

```python
trainf = pd.read_csv("/Users/vaibhavrangan/Downloads/Stat_303-3/Datasets/Car_features_train.csv")
trainp = pd.read_csv("/Users/vaibhavrangan/Downloads/Stat_303-3/Datasets/Car_prices_train.csv")
train = pd.merge(trainf, trainp)

testf = pd.read_csv("/Users/vaibhavrangan/Downloads/Stat_303-3/Datasets/Car_prices_test.csv")
testp = pd.read_csv("/Users/vaibhavrangan/Downloads/Stat_303-3/Datasets/Car_features_test.csv")
test = pd.merge(testf, testp)

predictors = ['mpg', 'engineSize', 'year', 'mileage']
target = 'price'
X_train = train[predictors]
y_train = train[target]
X_test = test[predictors]
y_test = test[target]
```

- no scaling, tree based model
- CatBoost handles categorical features more efficiently and in a more informative way.
- It is difficult to overfit – even with default input values


- each tree in the CatBoost model is trained on a subset of the training data, and its cost function is calculated on a DIFFERENT subset of the training data

```python
model = CatBoostRegressor(
    random_state = 12,

    # four gradient boosting hyperparameters are still present
    n_estimators = 20, 
    learning_rate = 0.1,
    max_depth = 4,
    subsample = 0.8,

    # still hae\ve reg_lambda to regularize the cost function
    reg_lambda = 1,
    verbose = False,

    thread_count = 1, # if we are using Quest
)

# no gamma, no colsample_ inputs: the extension in the theory above is enough to avoid overfitting

model.fit(X_train, y_train)
```

- feature_importances__ works the same way (only not normalized) as LightGBM
- the grid works the same way, except without gamma, colsample_ inputs
- for CatBoostClassifiers, scale_pos_weight is still available
